# Business Questions

## Analytical Questions

1. Which films are rented most frequently?
2. Which films generate the highest revenue?
3. Which film categories are most popular?
4. Which stores generate the highest number of rentals?
5. Which stores generate the highest revenue?
6. Which customers rent the most films?
7. Which customers generate the highest payments?
8. How does rental activity change over time?
9. How does revenue change by month, quarter, and year?
10. Which staff members process the highest number of rentals or payments?
11. Which cities or countries generate the highest rental activity?
12. Which films are returned late most often?

## Importance

These business questions help management identify profitable films, customer behavior patterns, store performance, revenue trends, and operational efficiency. This supports better decision-making related to inventory, customer retention, staffing, and revenue growth.
