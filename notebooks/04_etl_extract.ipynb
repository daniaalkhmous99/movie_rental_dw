# ETL Design - Extract Phase

## Source Tables Extracted from OLTP Database

The following tables are extracted from the operational movie rental database:

1. rental  
   - Used to analyze rental transactions and customer activity.

2. payment  
   - Used to analyze revenue and payment behavior.

3. customer  
   - Provides customer details for customer dimension analysis.

4. film  
   - Provides film information such as title, rental rate, and duration.

5. inventory  
   - Connects films to stores and available stock.

6. store  
   - Used for store performance analysis.

7. staff  
   - Used to analyze employee performance.

8. address  
   - Provides location details.

9. city  
   - Used for city-level analysis.

10. country  
   - Used for country-level analysis.

11. category  
   - Defines film categories.

12. film_category  
   - Maps films to categories.

13. language  
   - Provides film language information.

14. actor  
   - Stores actor information.

15. film_actor  
   - Maps films to actors.

## Reason for Extraction

These tables contain the transactional and descriptive data needed to populate fact tables and dimension tables in the data warehouse.

```python
# Extract all source tables
tables = ['rental', 'payment', 'customer', 'film', 
          'inventory', 'store', 'staff', 'address',
          'city', 'country', 'category', 'film_category',
          'language', 'actor', 'film_actor']

print("\nExtracted Tables:")
print("-" * 35)
for table in tables:
    cursor.execute(f"SELECT COUNT(*) FROM {table}")
    count = cursor.fetchone()[0]
    print(f"  ✔ {table:20} {count:,} rows")
```
