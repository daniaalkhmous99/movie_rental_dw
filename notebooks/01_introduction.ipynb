# Introduction

The purpose of this assignment is to transform an OLTP (Online Transaction Processing) movie rental database into a high-level data warehouse design suitable for analytical reporting and business intelligence.

The original OLTP database is designed for daily operational tasks such as managing customers, rentals, payments, staff, stores, and inventory. It is optimized for fast inserts, updates, and transactional consistency.

However, OLTP systems are not ideal for analytical queries because they contain highly normalized schemas and are optimized for transaction processing rather than reporting.

To support business decision-making, a data warehouse is required. A data warehouse uses OLAP concepts and dimensional modeling to organize data into fact tables and dimension tables.

This project proposes a dimensional model and ETL process to transform the operational movie rental system into an analytical data warehouse.
